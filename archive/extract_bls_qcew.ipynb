```python
from pathlib import Path
import pandas as pd

# Paths
input_dir = Path("../data/raw/QCEW_NAICS-based")
output_file = Path("../data/raw/qcew_usvi_accommodation_food_services_2000_2025.csv")

# Filters
columns = [
    "area_fips",
    "own_code",
    "industry_code",
    "size_code",
    "year",
    "annual_avg_estabs",
    "annual_avg_emplvl",
    "total_annual_wages",
    "annual_avg_wkly_wage",
    "avg_annual_pay",
]

industries = ["72", "721", "722"]

# Remove previous output, if present
output_file.unlink(missing_ok=True)

files = sorted(input_dir.glob("*.annual.singlefile.csv"))

if not files:
    raise FileNotFoundError(f"No annual CSV files found in {input_dir}")

for i, file in enumerate(files):
    print(f"Processing {file.name}...")

    df = pd.read_csv(
        file,
        usecols=columns,
        dtype={
            "area_fips": str,
            "own_code": str,
            "industry_code": str,
            "size_code": str,
        },
        low_memory=False,
    )

    filtered = df.loc[
        (df["area_fips"] == "78000")
        & (df["own_code"] == "5")
        & (df["industry_code"].isin(industries))
        & (df["size_code"] == "0"),
        columns,
    ]

    filtered.to_csv(
        output_file,
        mode="w" if i == 0 else "a",
        header=(i == 0),
        index=False,
    )

    print(f"  Extracted {len(filtered)} rows")

print(f"\nDone. Output: {output_file}")
```
